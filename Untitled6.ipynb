```python
train_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    '/content/Tien_VN/Tien_VN',  # Assuming a folder structure with a "train" directory
    target_size=(64, 64),  # Resize images
    batch_size=32,
    class_mode='categorical'  # Or 'binary' for binary classification
)

test_generator = test_datagen.flow_from_directory(
    '/content/Tien_VN/Tien_VN',  # Removed the extra space at the end of the path
    target_size=(64, 64),
    batch_size=32,
    class_mode='categorical'  # Or 'binary'
)

# 2. **Build Your ANN Model**
# Assuming you have 3 classes, adjust the final Dense layer accordingly
num_classes = train_generator.num_classes  # Get the number of classes from the generator
model = Sequential([
    Flatten(input_shape=(64, 64, 3)),  # Flatten the image input
    Dense(128, activation='relu'),  # Hidden layer with 128 neurons
    Dense(64, activation='relu'),   # Hidden layer with 64 neurons
    Dense(num_classes, activation='softmax')  # Output layer with num_classes neurons
])

# 3. **Compile the Model**
model.compile(optimizer=Adam(),
              loss='categorical_crossentropy',  # Or 'binary_crossentropy' for binary classification
              metrics=['accuracy'])

# 4. **Train the Model**
model.fit(train_generator, epochs=15, validation_data=test_generator)

# 5. **Evaluate the Model**
test_loss, test_acc = model.evaluate(test_generator)
print(f'Test Accuracy: {test_acc}')
```

```python
from keras.preprocessing.image import load_img, img_to_array
import numpy as np
import matplotlib.pyplot as plt

# Tải hình ảnh và chuẩn hóa về kích thước 64x64 (phù hợp với kích thước huấn luyện)
img = load_img('/content/z6505451415649_88b9bc31975c4093630da0138ee75b2f.jpg', target_size=(64, 64))

# Hiển thị hình ảnh
plt.imshow(img)
plt.axis('off')  # Tắt trục
plt.show()

# Chuyển đổi hình ảnh thành mảng numpy và chuẩn hóa
img_array = img_to_array(img) / 255.0  # Chuẩn hóa pixel về [0, 1]
img_flatten = img_array.reshape(1, 64, 64, 3)  # Đảm bảo hình dạng phù hợp với mô hình (64x64)

# Dự đoán
prediction = model.predict(img_flatten)

# In ra kết quả dự đoán
predicted_class = np.argmax(prediction)  # Lấy chỉ mục class có xác suất cao nhất
print('Tiên đoán là class:', predicted_class)
```
